# Baseline

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.metrics import classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
```

Read dataset files

```python
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

```python
train_df.info()
```

```python
train_df.Class.value_counts()
```

Make dataset

```python
train_df = train_df.drop(columns=['case_id', 'Hospital', 'patientid', 'City_Code_Patient', 'Patient_Visitors'], axis=1)
train_df = train_df.dropna()
X_train = train_df.iloc[:, :-1]
y_train = train_df.iloc[:, -1]
X_test = test_df.drop(columns=['case_id', 'Hospital', 'patientid', 'City_Code_Patient', 'Patient_Visitors'], axis=1)


print(len(X_train), len(X_test), len(y_train))
```

```python
train_df.head()
```

```python
from sklearn.model_selection import train_test_split

X = pd.concat([X_train, X_test], axis=0)
X = pd.get_dummies(X, drop_first=True)

X_train = X.iloc[:len(X_train)]
X_test = X.iloc[len(X_train):]
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.1, random_state=0)
```

```python
X_train.head()
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = { 'n_estimators' : [100],
           'max_depth' : [12],
           'min_samples_leaf' : [4, 8,10],
           'min_samples_split' : [4, 8, 10]
            }

estimator = RandomForestClassifier()

grid_search = GridSearchCV(estimator=estimator,
                          param_grid=param_grid,
                          cv=5,
                          scoring='f1_weighted',
                          n_jobs=-1)

grid_search.fit(X_train, y_train)
grid_search.best_params_
```

```python
{'max_depth': 12,
 'min_samples_leaf': 10,
 'min_samples_split': 8,
 'n_estimators': 100}
```

```python
param_grid = {'C': [0.001, 0.01, 0.1, 1, 10, 100],
              'penalty': ['l1', 'l2']}


grid_search = GridSearchCV(LogisticRegression(max_iter=1000), param_grid, cv=5,
                          n_jobs=-1)
grid_search.fit(X_train, y_train)
grid_search.score(X_test, y_test)
```

Make model & train

```python
model = RandomForestClassifier(n_estimators=100,
                               bootstrap=True,
                               oob_score=False,
                               max_samples=None,
                               random_state=0, 
                              max_depth=12, 
                              min_samples_leaf=10,
                              min_samples_split=8)
model.fit(X_train, y_train)

val_pred = model.predict(X_val)
print(classification_report(y_val, val_pred))
```

```python
c1f1 = LogisticRegression(multi_class='multinomial', random_state=1)
c1f2 = RandomForestClassifier(n_estimators=100,
                               bootstrap=True,
                               oob_score=False,
                               max_samples=None,
                               random_state=0, 
                              max_depth=12, 
                              min_samples_leaf=10,
                              min_samples_split=8)
c1f3 = GaussianNB()
ec1f1 = VotingClassifier(estimators=[
        ('lr', c1f1), ('rf', c1f2), ('gnb', c1f3)], voting='soft')
ec1f1 = ec1f1.fit(X_train, y_train)

val_pred = ec1f1.predict(X_val)
print(classification_report(y_val, val_pred))
```

Make test prediction

```python
test_pred = model.predict(X_test)
```

Export prediction to file

```python
submission_df = pd.read_csv('sample_submission.csv')
submission_df.Class = test_pred
submission_df.to_csv('submission.csv', index=False)
```
